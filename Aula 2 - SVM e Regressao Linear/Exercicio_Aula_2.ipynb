```python
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn import metrics
```

```python
#fazendo o download dos dados direto dos datasets presente na scikit-learn
data = datasets.load_breast_cancer()
data.feature_names
```

```python
#variável meta (target) -> aquilos que estamos tentando prever
#no caso: cancer benigno ou maligno
data.target_names
```

```python
#linhas x colunas
data.data.shape
```

```python
#visualizando as 5 primeiras amostras
data.data[0:5]
```

## TODO 1
> Separe o dataset em conjunto de treino e teste, sendo 70% para treino e 30% para teste

```python
#resposta

X = data.data    # features (30 colunas)
y = data.target  # target (0=maligno, 1=benigno)

split_teste_size = 0.30

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=split_teste_size , random_state=42
)

print(f"Treino: {X_train.shape[0]} amostras")
print(f"Teste:  {X_test.shape[0]} amostras")
```

## TODO 2
> Treine um classificador SVM usando kernel linear e outro usando kernel RBF

```python
#resposta

# Kernel Linear
clf_linear = svm.SVC(kernel='linear')
clf_linear.fit(X_train, y_train)

# Kernel RBF (Radial Basis Function)
clf_rbf = svm.SVC(kernel='rbf')
clf_rbf.fit(X_train, y_train)
```

## TODO 3
> Faça a predição para o conjunto de teste usando os dois kernels

```python
#resposta

y_pred_linear = clf_linear.predict(X_test)
y_pred_rbf    = clf_rbf.predict(X_test)
```

## TODO 4
> Obtenha a Acurácia, Precision e Recall para cada kernel

```python
#resposta

for nome, y_pred in [("Linear", y_pred_linear), ("RBF", y_pred_rbf)]:
    acc  = metrics.accuracy_score(y_test, y_pred)
    prec = metrics.precision_score(y_test, y_pred)
    rec  = metrics.recall_score(y_test, y_pred)
    print(f"--- Kernel {nome} ---")
    print(f"Acurácia:  {acc:.4f}")
    print(f"Precision: {prec:.4f}")
    print(f"Recall:    {rec:.4f}\n")
```

## TODO 5
> Você consegue explicar o motivos dos indicadores serem diferentes?

#resposta

Essa é a pergunta mais importante do exercício! Aqui está a explicação:
Kernel Linear traça um hiperplano reto para separar as classes. Funciona muito bem quando os dados são linearmente separáveis — e features médicas como as do breast cancer tendem a ter boa separação linear.
Kernel RBF (Radial Basis Function) projeta os dados em um espaço de dimensão maior para encontrar fronteiras curvas e não-lineares. É mais flexível, mas pode sofrer de overfitting se os hiperparâmetros C e gamma não estiverem ajustados.
Por isso você pode ver que:

O Linear costuma ter Recall mais alto nesse dataset — erra menos tumores malignos reais
O RBF sem tuning pode ter métricas piores porque usa valores padrão de C e gamma que não são ideais para a escala dos dados


💡 Dica importante: no breast cancer, Recall é a métrica mais crítica — um falso negativo (dizer que é benigno quando é maligno) é muito mais perigoso que um falso positivo. Por isso não basta olhar só a acurácia!

O código abaixo prepara os dados para executar uma regressao linear

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
lemonades = pd.read_csv('bases/Lemonades.csv', sep = ";")
lemonades = lemonades.drop_duplicates()
lemonades['Date'] = pd.to_datetime(lemonades['Date'])
t0 = lemonades.Date.min()
m = lemonades.Date.notnull()
lemonades.loc[m, 'Date_int'] = (lemonades.loc[m, 'Date'] - t0).dt.total_seconds()
lemonades['Date'] = np.datetime64(t0) + pd.to_timedelta(lemonades.Date_int.interpolate(), unit='s')
del lemonades['Date_int']
lemonades['Leaflets'].fillna((lemonades['Leaflets'].mean()), inplace=True)
lemonades["Sales"] = lemonades["Lemon"] + lemonades["Orange"]
lemonades["Revenue"] = lemonades["Sales"] * lemonades["Price"]
```

Ajuste uma Regressão Linear aos dados para predizer vendas usando cada uma das features abaixo:

* temperature
* leaflets
* price

Calcule o RMSE

```python
# temperature

X_temp = lemonades[['Temperature']]
y = lemonades['Sales']

model_temp = LinearRegression()
model_temp.fit(X_temp, y)
y_pred_temp = model_temp.predict(X_temp)

rmse_temp = np.sqrt(mean_squared_error(y, y_pred_temp))
print(f"RMSE - Temperature: {rmse_temp:.4f}")
```

```python
# leaflets

# ❌ forma antiga (depreciada)
#lemonades['Leaflets'].fillna((lemonades['Leaflets'].mean()), inplace=True)

# ✅ forma correta
lemonades['Leaflets'] = lemonades['Leaflets'].fillna(lemonades['Leaflets'].mean())


X_leaf = lemonades[['Leaflets']]

model_leaf = LinearRegression()
model_leaf.fit(X_leaf, y)
y_pred_leaf = model_leaf.predict(X_leaf)

rmse_leaf = np.sqrt(mean_squared_error(y, y_pred_leaf))
print(f"RMSE - Leaflets: {rmse_leaf:.4f}")
```

```python
# price
X_price = lemonades[['Price']]

model_price = LinearRegression()
model_price.fit(X_price, y)
y_pred_price = model_price.predict(X_price)

rmse_price = np.sqrt(mean_squared_error(y, y_pred_price))
print(f"RMSE - Price: {rmse_price:.4f}")
```
